## 가설 2 & 3: 버스 이동 및 도착 시간 편차 분석

버스의 운행 정시성을 파악하기 위해 두 가지를 분석합니다.
1.  **버스 이동 편차**: 각 버스가 정류장 사이를 이동하는 데 걸리는 시간의 편차를 분석합니다.
2.  **버스 도착 편차**: 실제 버스 도착 시간과 공식 시간표의 차이를 분석합니다.

이를 위해 먼저 각 운행별로 정류장 간 이동 시간을 계산하고, 이를 정류장별로 그룹화하여 시간 편차를 박스 플롯으로 시각화합니다.
박스 플롯은 데이터의 중앙값, 사분위수, 이상치 등을 한눈에 보여주어 분포를 파악하는 데 유용합니다.

```python
# 'bstop_arrive_time'을 다시 날짜/시간 객체로 변환하여 시간 계산을 준비합니다.
_gs_busevent_preprocessed818['arrive_time_dt'] = pd.to_datetime(
    _gs_busevent_preprocessed818['bstop_arrive_time'].astype(str), 
    format='%Y%m%d%H%M%S'
)

# 각 버스(vehicle_id)별로 도착 시간에 따라 데이터를 정렬합니다.
# [수정] 'plate_no' 대신 버스 차량을 고유하게 식별하는 컬럼명('vehicle_id'로 추정)을 사용합니다.
# 만약 다른 이름이라면, 위쪽 디버깅 셀에서 확인된 실제 컬럼명으로 수정해야 합니다.
sorted_bus_data = _gs_busevent_preprocessed818.sort_values(by=['route_name', 'arrive_time_dt'])

# 각 버스별로 이전 정류장에서 현재 정류장까지의 이동 시간을 계산합니다.
# diff()는 현재 행과 이전 행의 차이를 계산해줍니다.
sorted_bus_data['travel_time'] = sorted_bus_data.groupby('route_name')['arrive_time_dt'].diff().dt.total_seconds() / 60.0

# 이동 시간이 0보다 크고 30분 미만인 유효한 데이터만 필터링합니다. (이상치 제거)
valid_travel_time = sorted_bus_data[(sorted_bus_data['travel_time'] > 0) & (sorted_bus_data['travel_time'] < 30)]

# 차트 크기 설정
plt.figure(figsize=(20, 10))

# 정류장별 이동 시간 분포를 박스 플롯으로 시각화합니다.
sns.boxplot(x='node_name', y='travel_time', data=valid_travel_time)

# 차트 제목과 축 레이블 설정
plt.title('주요 정류장별 버스 이동 시간 편차 (818번 버스)')
plt.xlabel('버스 정류장')
plt.ylabel('이동 시간 (분)')
plt.xticks(rotation=90) # 정류장 이름이 길기 때문에 90도 회전

# 차트 출력
plt.show()
```

## 가설 4: 버스 시간표와 실제 도착 시간 편차 분석 (회차별)

공식 버스 시간표 데이터와 실제 운행 기록을 병합하여, 각 버스 회차별로 시간표 대비 실제 도착 시간이 얼마나 차이 나는지 분석합니다.
이를 통해 특정 회차의 버스가 자주 지연되는지, 혹은 너무 일찍 도착하는지 등의 패턴을 파악할 수 있습니다.

요청에 따라, (실제 도착 시간 - 시간표 상 도착 시간) + 10분의 값을 계산하여 시각화합니다.
이 분석을 위해 두 데이터를 '회차'와 '정류장' 기준으로 병합하는 전처리 과정이 필요합니다.

```python
# 시간표 데이터(_bs_node)의 컬럼 이름을 분석에 용이하도록 변경합니다.
_bs_node.columns = ['round_trip', 'jinryang_start', 'daegu_univ1', 'hayang_station1', 'chamber_of_commerce', 
                   'ansim_station', 'hayang_station2', 'daegu_univ2', 'jinryang_end', 'notes']

# 'round_trip' (회차)를 기준으로 시간표 데이터를 긴 형태로 변환합니다.
# 각 정류장 이름이 'node_name' 열에, 시간표 시간이 'scheduled_time' 열에 오게 됩니다.
timetable_long = _bs_node.melt(id_vars=['round_trip'], 
                               value_vars=['jinryang_start', 'daegu_univ1', 'hayang_station1', 'chamber_of_commerce', 
                                           'ansim_station', 'hayang_station2', 'daegu_univ2', 'jinryang_end'],
                               var_name='bstop_name_code', value_name='scheduled_time')

# 정류장 코드(bstop_name_code)를 실제 정류장 이름으로 매핑합니다. (실제 데이터의 정류장 이름과 일치시켜야 함)
# 이 부분은 실제 데이터의 정류장 이름에 맞게 수정이 필요할 수 있습니다.
bstop_name_mapping = {
    'jinryang_start': '진량황제A', 
    'daegu_univ1': '대구대',
    'hayang_station1': '하양역',
    'chamber_of_commerce': '영천상공회의소',
    'ansim_station': '안심역',
    'hayang_station2': '하양역',
    'daegu_univ2': '대구대',
    'jinryang_end': '진량황제A'
}
timetable_long['node_name'] = timetable_long['bstop_name_code'].map(bstop_name_mapping)

# 실제 운행 데이터와 시간표를 병합하기 위해, 날짜와 시간을 합쳐 완전한 날짜/시간 객체를 생성합니다.
# 시간표에는 날짜 정보가 없으므로, 실제 운행 데이터의 날짜를 가져와 사용합니다.
merged_data = pd.merge(
    _gs_busevent_preprocessed818,
    timetable_long,
    on='node_name',
    how='inner'
)

# 시간표의 시간(HH:MM)과 운행 데이터의 날짜를 결합합니다.
merged_data['scheduled_datetime'] = pd.to_datetime(
    merged_data['bstop_arrive_days'] + ' ' + merged_data['scheduled_time'],
    errors='coerce' # 변환 중 오류가 발생하면 NaT(Not a Time)으로 처리
)

# 유효한 날짜/시간 데이터만 필터링합니다.
merged_data = merged_data.dropna(subset=['scheduled_datetime', 'arrive_time_dt'])

# 실제 도착 시간과 시간표 상의 시간 차이를 분 단위로 계산합니다.
merged_data['time_diff'] = (merged_data['arrive_time_dt'] - merged_data['scheduled_datetime']).dt.total_seconds() / 60.0

# 요청에 따라 시간 차이에 10분을 더합니다.
merged_data['adjusted_diff'] = merged_data['time_diff'] + 5

# 회차(round_trip)별로 조정된 시간 편차의 평균을 계산합니다.
avg_diff_by_roundtrip = merged_data.groupby('round_trip')['adjusted_diff'].mean()

# 차트 크기 설정
plt.figure(figsize=(20, 10))

# 회차별 평균 도착 시간 편차를 막대 차트로 시각화합니다.
sns.barplot(x=avg_diff_by_roundtrip.index, y=avg_diff_by_roundtrip.values)

# 차트 제목과 축 레이블 설정
plt.title('버스 회차별 시간표 대비 도착 시간 편차 (+10분 조정)')
plt.xlabel('버스 회차 순번')
plt.ylabel('평균 도착 시간 편차 (분)')
plt.xticks(rotation=90)

# 차트 출력
plt.show()
```
